# 04 · HMM Regime Detection

Fit a Gaussian HMM, map states to regimes, and visualise transitions.


```python
import sys, warnings
warnings.filterwarnings('ignore')
sys.path.append('..')  # import the src package from the repo root
from src.utils.config import load_config
from src.data import MarketDataLoader
from src.cleaning import DataCleaner
cfg = load_config(env='research')
TICKER = cfg.data.tickers[0]
from src.regime import HMMRegimeDetector
from src.reporting import plot_regimes
df = DataCleaner(cfg).clean(MarketDataLoader(cfg).load_one(TICKER))
detector = HMMRegimeDetector(cfg).fit(df)
regimes = detector.predict(df)
regimes['regime'].value_counts()
```

## State statistics (mean return / vol per latent state)


```python
import pandas as pd; pd.DataFrame(detector.state.state_stats).T
```

## Regime overlay on price


```python
plot_regimes(df['close'], regimes['regime'], out='../reports/regimes.png')
```
