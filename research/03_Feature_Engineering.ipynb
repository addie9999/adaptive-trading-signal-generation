# 03 · Feature Engineering

Build ~50 features via reusable generator classes and inspect them.


```python
import sys, warnings
warnings.filterwarnings('ignore')
sys.path.append('..')  # import the src package from the repo root
from src.utils.config import load_config
from src.data import MarketDataLoader
from src.cleaning import DataCleaner
cfg = load_config(env='research')
TICKER = cfg.data.tickers[0]
from src.features import FeatureEngineer
df = DataCleaner(cfg).clean(MarketDataLoader(cfg).load_one(TICKER))
eng = FeatureEngineer(cfg)
X, y = eng.build_dataset(df)
print('features:', X.shape[1]); X.iloc[-3:, :8]
```

## Feature correlation with the forward-return label


```python
corr = X.apply(lambda c: c.corr(y)).abs().sort_values(ascending=False)
corr.head(15)
```
