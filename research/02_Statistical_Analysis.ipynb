# 02 · Statistical Analysis

Stationarity, autocorrelation, normality and the Hurst exponent.


```python
import sys, warnings
warnings.filterwarnings('ignore')
sys.path.append('..')  # import the src package from the repo root
from src.utils.config import load_config
from src.data import MarketDataLoader
from src.cleaning import DataCleaner
cfg = load_config(env='research')
TICKER = cfg.data.tickers[0]
from src.statistics import adf_test, kpss_test, ljung_box_test, normality_test, hurst_exponent
df = DataCleaner(cfg).clean(MarketDataLoader(cfg).load_one(TICKER))
rets = df['close'].pct_change().dropna()
```

## Unit-root / stationarity


```python
print(adf_test(df['close']))   # price level: expect non-stationary
print(adf_test(rets))          # returns: expect stationary
print(kpss_test(rets))
```

## Autocorrelation & normality


```python
print(ljung_box_test(rets)); print(normality_test(rets))
```

## Hurst exponent (mean-reversion vs trend)


```python
print('Hurst:', round(hurst_exponent(df['close'].apply('log')), 3))
```
