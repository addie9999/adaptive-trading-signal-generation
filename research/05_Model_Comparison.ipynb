# 05 · Multi-Model Comparison

Train ARIMA / XGBoost (and optionally deep models) and compare metrics.


```python
import sys, warnings
warnings.filterwarnings('ignore')
sys.path.append('..')  # import the src package from the repo root
from src.utils.config import load_config
from src.data import MarketDataLoader
from src.cleaning import DataCleaner
cfg = load_config(env='research')
TICKER = cfg.data.tickers[0]
from src.features import FeatureEngineer
from src.models import build_model
df = DataCleaner(cfg).clean(MarketDataLoader(cfg).load_one(TICKER))
X, y = FeatureEngineer(cfg).build_dataset(df)
split = int(len(X)*0.8)
rows = []
for name in ['arima','xgboost']:
    m = build_model(name, cfg).fit(X[:split], y[:split])
    rows.append({'model': name, **m.evaluate(X[split:], y[split:])})
import pandas as pd; pd.DataFrame(rows)
```
